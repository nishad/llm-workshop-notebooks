## 00.01: What is a Large Language Model (LLM)? ⏱️ 5 minutes

### Understanding LLMs in Plain English

**Large Language Models (LLMs)** are AI systems trained on massive amounts of text data to understand and generate human-like language. Think of them as sophisticated pattern-recognition systems that have "read" billions of pages of text and learned statistical relationships between words, concepts, and ideas.

### How Do LLMs Work?

1. **Training Phase**: The model reads enormous amounts of text (books, websites, articles) and learns patterns
2. **Pattern Recognition**: It learns which words tend to appear together, how sentences are structured, and how concepts relate
3. **Generation**: When you give it a prompt, it predicts the most likely next words based on patterns it learned

### Key Concept: It's Not Magic, It's Math

LLMs don't "understand" in the human sense—they're excellent at predicting what text should come next based on patterns. But this prediction is so sophisticated that it often appears like understanding!

### Architecture Basics

Modern LLMs use an architecture called **Transformers**:

```
Your Prompt → [Tokenization] → [Transformer Layers] → [Prediction] → Response
```

- **Tokenization**: Text is broken into smaller pieces (tokens)
- **Transformer Layers**: Neural network layers that process the tokens and identify patterns
- **Prediction**: The model predicts the most likely next token, repeatedly, to generate a response

### Model Size & Parameters

When you see "3B" or "7B" in model names:
- **3B = 3 billion parameters** (connections in the neural network)
- **7B = 7 billion parameters**

**General Rule**: 
- More parameters → Better quality, but slower and more memory-intensive
- Fewer parameters → Faster, less memory, but potentially lower quality

### Analogy

Think of an LLM like a very well-read assistant who has read millions of books but can only work with the patterns they've seen during their "reading" (training). They're great at synthesizing information from those patterns but don't have access to information beyond their training cutoff.

## 00.02: Local vs. Cloud LLMs ⏱️ 5 minutes

### Why Choose Local LLMs?

| Aspect | Cloud LLMs (ChatGPT, Claude) | Local LLMs (Ollama) |
|--------|------------------------------|---------------------|
| **Privacy** | Data sent to external servers | Data stays on your machine |
| **Cost** | Subscription or pay-per-use | Free after initial setup |
| **Internet** | Required | Works offline |
| **Speed** | Depends on network | Depends on your hardware |
| **Quality** | Often higher (larger models) | Good for many tasks |
| **Customization** | Limited | Full control over models |
| **Compliance** | May have restrictions | Fully compliant for sensitive data |

### When to Use Local LLMs

✅ **Perfect for:**
- Sensitive data (medical, legal, financial)
- Offline environments
- High-volume usage (avoid API costs)
- Learning and experimentation
- Full control and customization needs

⚠️ **Consider cloud alternatives for:**
- Tasks requiring the absolute best quality
- When you don't have suitable hardware
- Quick prototyping without setup

### What Makes 2025 Special?

Recent developments have made local LLMs much more accessible:
1. **Smaller, smarter models**: 3B parameter models now rival older 7B models
2. **Better tools**: Ollama makes setup as easy as installing an app
3. **Optimized for consumer hardware**: Models that run well on standard laptops
4. **Improved quantization**: Compress models without significant quality loss

````python
# 00.03.2: Check Available Memory and Disk Space
import psutil
import shutil

# Check RAM
memory = psutil.virtual_memory()
total_ram_gb = memory.total / (1024**3)
available_ram_gb = memory.available / (1024**3)

print("\n" + "=" * 50)
print("MEMORY CHECK")
print("=" * 50)
print(f"Total RAM: {total_ram_gb:.2f} GB")
print(f"Available RAM: {available_ram_gb:.2f} GB")

if total_ram_gb >= 8:
    print("✅ Sufficient RAM for workshop")
    if total_ram_gb >= 16:
        print("   💪 Excellent! You can run larger models")
else:
    print("⚠️  RAM is below recommended 8GB. You may experience slowness.")

# Check disk space
disk = shutil.disk_usage(".")
free_space_gb = disk.free / (1024**3)

print("\n" + "=" * 50)
print("DISK SPACE CHECK")
print("=" * 50)
print(f"Free disk space: {free_space_gb:.2f} GB")

if free_space_gb >= 20:
    print("✅ Sufficient disk space")
else:
    print("⚠️  Less than 20GB free. Please free up space before downloading models.")
````

````python
# 00.03.3: Check if Ollama is Installed and Running
import subprocess

def check_ollama_status():
    """
    Check if Ollama is installed and running.
    Returns: (installed: bool, running: bool)
    """
    try:
        # Try to run 'ollama list' command
        result = subprocess.run(
            ['ollama', 'list'],
            capture_output=True,
            timeout=5,
            text=True
        )
        return True, result.returncode == 0
    except FileNotFoundError:
        return False, False
    except subprocess.TimeoutExpired:
        return True, False
    except Exception:
        return False, False

print("\n" + "=" * 50)
print("OLLAMA CHECK")
print("=" * 50)

installed, running = check_ollama_status()

if installed and running:
    print("✅ Ollama is installed and running!")
    print("   You're all set to continue.")
elif installed and not running:
    print("⚠️  Ollama is installed but not running.")
    print("   Please start Ollama:")
    print("   - macOS/Linux: Run 'ollama serve' in a terminal")
    print("   - Windows: Start the Ollama application")
else:
    print("❌ Ollama is not installed.")
    print("\n📥 Installation Instructions:")
    print("   • macOS:   brew install ollama")
    print("   • Linux:   curl https://ollama.ai/install.sh | sh")
    print("   • Windows: Download from https://ollama.com/download")
    print("\n   After installation, run 'ollama serve' to start the service.")
````

## 00.05: Model Selection Guide ⏱️ 5 minutes

### Understanding Model Naming

Ollama models are named like: `model-name:size`

Examples:
- `llama3.2:1b` - Llama 3.2 with 1 billion parameters
- `llama3.2:3b` - Llama 3.2 with 3 billion parameters
- `mistral:7b` - Mistral with 7 billion parameters

### Choosing the Right Model

| Model | Size | RAM Needed | Speed | Quality | Best For |
|-------|------|------------|-------|---------|----------|
| **llama3.2:1b** | ~1.3GB | 4GB+ | ⚡⚡⚡ | ⭐⭐ | Quick tests, learning, low-end hardware |
| **llama3.2:3b** | ~2GB | 8GB+ | ⚡⚡ | ⭐⭐⭐ | **Workshop choice** - Good balance |
| **llama3.2:7b** | ~4.7GB | 16GB+ | ⚡ | ⭐⭐⭐⭐ | Better quality, if you have RAM |
| **mistral:7b** | ~4.1GB | 16GB+ | ⚡ | ⭐⭐⭐⭐ | Alternative to Llama |

### For This Workshop

We'll use **`llama3.2:3b`** because it:
- ✅ Runs on 8GB RAM systems
- ✅ Provides good quality responses
- ✅ Downloads quickly (~2GB)
- ✅ Responds fast enough for interactive use

### Specialized Models

We'll also use:
- **`nomic-embed-text`** - For generating embeddings (Part 2: RAG)
  - Size: ~274MB
  - Purpose: Convert text to vector representations

### Model Quantization

**Q: What does "quantization" mean?**

A: It's a compression technique that reduces model size with minimal quality loss.
- Models in Ollama are typically 4-bit quantized
- Original model might be 12GB, quantized version is 3GB
- Quality difference is usually minimal for most tasks

## 00.06: Downloading Models ⏱️ 5-10 minutes

Now let's download the models we'll need for the workshop.

**⏱️ Time Estimate:**
- Fast connection (50+ Mbps): 3-5 minutes
- Medium connection (10-50 Mbps): 5-10 minutes
- Slow connection (<10 Mbps): 10-20 minutes

**Total Download Size**: ~2.3GB

````python
# 00.06.1: Check what models are already installed
import ollama

print("=" * 50)
print("CURRENTLY INSTALLED MODELS")
print("=" * 50)

try:
    models = ollama.list()

    if models.models:  # Changed from models.get('models')
        for model in models.models:  # Changed from models['models']
            name = model.model  # Changed from model.get('name', 'Unknown')
            size_bytes = model.size  # Changed from model.get('size', 0)
            size_gb = size_bytes / (1024**3)
            print(f"  • {name:30s} ({size_gb:.2f} GB)")
        print(f"\nTotal models: {len(models.models)}")
    else:
        print("  No models installed yet.")

except Exception as e:
    print(f"❌ Error connecting to Ollama: {e}")
    print("\nMake sure Ollama is running:")
    print("  Run 'ollama serve' in a terminal")

print("=" * 50)
````

````python
# 00.06.2: Download required models
import ollama
from datetime import datetime

def download_model_if_needed(model_name):
    """
    Check if a model exists, and download it if it doesn't.
    """
    print(f"\n{'='*50}")
    print(f"Checking: {model_name}")
    print('='*50)
    
    try:
        # Try to get model info (this works if model exists)
        ollama.show(model_name)
        print(f"✅ {model_name} is already installed!")
        return True
        
    except:
        # Model doesn't exist, download it
        print(f"📥 Downloading {model_name}...")
        print(f"   This may take several minutes depending on your connection.")
        print(f"   Started at: {datetime.now().strftime('%H:%M:%S')}")
        
        try:
            ollama.pull(model_name)
            print(f"\n✅ {model_name} downloaded successfully!")
            print(f"   Completed at: {datetime.now().strftime('%H:%M:%S')}")
            return True
        except Exception as e:
            print(f"\n❌ Error downloading {model_name}: {e}")
            return False

# Models we need for the workshop
required_models = [
    'llama3.2:3b',        # Main text generation model
    'nomic-embed-text'    # Embedding model for RAG
]

print("🎯 WORKSHOP MODEL SETUP")
print("\nWe need to download 2 models:")
print("  1. llama3.2:3b      (~2GB)  - Text generation")
print("  2. nomic-embed-text (~274MB) - Embeddings")
print("\nTotal: ~2.3GB")
print("\nStarting download process...")

results = {}
for model in required_models:
    results[model] = download_model_if_needed(model)

# Summary
print("\n" + "="*50)
print("DOWNLOAD SUMMARY")
print("="*50)

all_success = True
for model, success in results.items():
    status = "✅" if success else "❌"
    print(f"{status} {model}")
    if not success:
        all_success = False

if all_success:
    print("\n🎉 All models are ready! You're set for the workshop.")
else:
    print("\n⚠️  Some models failed to download.")
    print("   Please check your internet connection and try again.")
    print("   Or download manually: ollama pull <model-name>")
````

### 💡 What Just Happened?

When you download a model with Ollama:

1. **Model is fetched** from Ollama's registry (like Docker images)
2. **Stored locally** in Ollama's model directory:
   - macOS: `~/.ollama/models/`
   - Linux: `/usr/share/ollama/.ollama/models/`
   - Windows: `C:\Users\<username>\.ollama\models\`
3. **Ready to use** - No need to download again!

### 🎯 Manual Download (Alternative)

You can also download models from the terminal:

```bash
ollama pull llama3.2:3b
ollama pull nomic-embed-text
```

This is useful if you want to download models before the workshop!
